## Checking for Missing Values

```python
data.isnull().sum() #checking the columns with missing values
```

The columns with missing values are host_name,neighbourhood_group,price,last_review,reviews_per_month and license.

Remembering that the key focus of the project is predicting the price of an airbnb, with the price being our target. Since our data has 292,802 rows and 152,852 rows with null values in the price column, I decided to drop the null values in price, instead of filling the price target with measures of central tendencies, because our model would learn our 'imputed' prices instead of learning the actual price trends.

I also decided to drop the neighborhood_group,last_review,reviews_per_month and license columns as they don't have much significance to the price prediction. 

I also filled missing values in host_name with a placeholder 'Unknown'

#### Dropping columns
##### 'neighbourhood_group','last_review','reviews_per_month','license'

```python
data = data.drop(columns=['neighbourhood_group','last_review','reviews_per_month','license'])
#remove the above 4 columns
```

```python
data.shape #confirming that the 4 columns have been dropped
```

#### Dropping rows with missing values in the Price column

```python
data = data.dropna(subset=['price'])
data.isnull().sum() #confirming that the price column doesn't have any missing values
```

#### Filling missing values in host_name column with placeholder 'Unknown'

```python
data['host_name'] = data['host_name'].fillna('Unknown')
```

```python
data.isnull().sum() #confirming that there are no missing values in the data
```
